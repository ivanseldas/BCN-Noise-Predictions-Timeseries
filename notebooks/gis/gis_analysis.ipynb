```python
import sys, os
# Add the project root to Python path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '../..')))

import geopandas as gpd
from src.gis.network_features import load_graph, compute_betweenness, attach_centrality_to_sensors, export_edges_with_centrality
```

```python
# 1️⃣ Load the street network graph for Barcelona
G = load_graph("Barcelona, Spain", network_type="drive")
```

```python
import osmnx as ox
import geopandas as gpd
import folium

# Convert the network edges to a GeoDataFrame
edges = ox.graph_to_gdfs(G, nodes=False, edges=True)

# Create a folium map centered on Barcelona
m = folium.Map(location=[41.3851, 2.1734], zoom_start=12, tiles='OpenStreetMap')

# Add the network edges as GeoJSON
folium.GeoJson(
    edges.to_json(),
    name="Street Network",
    style_function=lambda x: {
        "color": "#FF5733",
        "weight": 1.5,
        "opacity": 0.5
    }
).add_to(m)

folium.LayerControl(collapsed=False).add_to(m)

m
```

```python
# 2️⃣ Calcular centralidad (muestra 200 nodos para hacerlo rápido)
centrality = compute_betweenness(G, k=200)
```

```python
# 3️⃣ Cargar tus sensores con coordenadas
import pandas as pd 
import geopandas as gpd

sensors = pd.read_csv("../../data/raw/sensors/XarxaSoroll_EquipsMonitor_Instal.csv") 
sensors_gdf = gpd.GeoDataFrame(sensors, geometry=gpd.points_from_xy(sensors.Longitud, sensors.Latitud), crs="EPSG:4326")
sensors_gdf["geometry"].head()
```

```python
# 4️⃣ Asociar la centralidad de nodo más cercano
sensors = attach_centrality_to_sensors(sensors_gdf, G, centrality)
sensors.head()
```

```python
# 5️⃣ Exportar la red vial con centralidad media de aristas
edges = export_edges_with_centrality(G, centrality, "../../data/gis_layers/barcelona_centrality.gpkg")
edges.head()
```

```python
# 6️⃣ Guardar los sensores enriquecidos
sensors.to_file("../../data/gis_layers/sensors_with_centrality.geojson", driver="GeoJSON")

sensors.head()
```
