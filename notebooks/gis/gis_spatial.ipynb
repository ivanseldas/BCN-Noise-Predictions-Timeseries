```python
import sys, os
# Add the project root to Python path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '../..')))

import geopandas as gpd
from src.gis.spatial_features import sjoin_nearest_landuse, distance_to_primary_roads, local_noise_mean
```

```python
# Load data
sensors = gpd.read_file("../../data/gis/gis_layers/sensors_with_centrality.geojson").to_crs(4326)
landuse = gpd.read_file("data/gis_layers/landuse.geojson").to_crs(4326)
edges = gpd.read_file("data/gis_layers/barcelona_centrality.gpkg").to_crs(4326)
```

```python
# 1️⃣ Nearest land use
sensors = sjoin_nearest_landuse(sensors, landuse, landuse_col="landuse")
```

```python
# 2️⃣ Distance to primary roads
sensors = distance_to_primary_roads(sensors, edges)
```

```python
# 3️⃣ Local mean noise
sensors = local_noise_mean(sensors, radius_m=150, value_col="noise_pred")
```

```python
# Save enriched sensors
sensors.to_file("data/gis_layers/sensors_gis_features.geojson", driver="GeoJSON")
sensors.head()
```
